# Matplotlib & Seaborn 기초

데이터 시각화는 ML 워크플로의 모든 단계에서 쓰입니다:

- **탐색(EDA)**: 데이터 분포, 이상치, 변수 간 관계 파악
- **모델링**: 학습 곡선으로 과적합/과소적합 진단
- **보고**: 결과를 다른 사람에게 전달

| 라이브러리 | 역할 |
|------------|------|
| **Matplotlib** | 저수준 기본 도구 — 모든 요소를 세밀하게 제어 |
| **Seaborn** | Matplotlib 위에 만든 고수준 도구 — 통계 차트를 예쁘게, 짧은 코드로 |

## 학습 목표
- Matplotlib의 기본 구조 (Figure / Axes) 이해
- 상황에 맞는 차트 고르기 (선, 막대, 히스토그램, 산점도, 박스플롯)
- 서브플롯으로 여러 차트 배치, 파일로 저장
- Seaborn으로 통계 시각화 (분포, 관계, 상관관계)
- ML 실무 시각화 (특성 분포, 학습 곡선, 이미지 격자)

## 이 노트북 사용법
각 섹션은 **개념 설명 → 예제 코드 → ✏️ 직접 해보기 → 정답** 순서입니다.

## 0. 왜 시각화가 필요할까?

숫자 요약(평균, 표준편차)만 보면 완전히 다른 데이터도 똑같아 보일 수 있습니다.
아래 두 데이터는 **평균과 표준편차가 거의 같지만** 분포 모양이 전혀 다릅니다.

````python
np.random.seed(42)

# A: 정규분포 / B: 두 봉우리(bimodal) 분포 — 평균과 표준편차를 비슷하게 맞춤
data_a = np.random.normal(50, 12.2, 1000)
data_b = np.concatenate([np.random.normal(38, 4, 500), np.random.normal(62, 4, 500)])

print(f"A: 평균 {data_a.mean():.1f}, 표준편차 {data_a.std():.1f}")
print(f"B: 평균 {data_b.mean():.1f}, 표준편차 {data_b.std():.1f}")
print("→ 숫자만 보면 거의 같은 데이터?")

fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].hist(data_a, bins=30, color='steelblue', edgecolor='white')
axes[0].set_title('데이터 A')
axes[1].hist(data_b, bins=30, color='indianred', edgecolor='white')
axes[1].set_title('데이터 B — 봉우리가 2개!')
plt.tight_layout()
plt.show()
````

## 1. Matplotlib 기본 구조

Matplotlib 그림은 두 계층으로 이루어집니다:

- **Figure**: 전체 도화지 (`plt.figure(figsize=(가로, 세로))`)
- **Axes**: 도화지 안의 실제 그래프 영역 (하나의 Figure에 여러 개 가능)

기본 차트 하나를 그리는 표준 패턴:

```python
plt.figure(figsize=(10, 5))   # 1. 도화지 준비
plt.plot(x, y, label='이름')   # 2. 그리기
plt.xlabel('x축 이름')          # 3. 꾸미기 (제목, 축, 범례, 격자)
plt.title('제목')
plt.legend()
plt.grid(True)
plt.show()                     # 4. 출력
```

````python
# 선 그래프 — 연속적인 변화(시간에 따른 값, 함수)를 표현
x = np.linspace(0, 10, 100)

plt.figure(figsize=(10, 5))
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)', linestyle='--')   # 선 스타일 변경
plt.xlabel('x')
plt.ylabel('y')
plt.title('사인과 코사인')
plt.legend()          # label을 지정했으면 legend()로 범례 표시
plt.grid(True, alpha=0.3)
plt.show()
````

````python
# 선 스타일 옵션 모음 — color, linestyle, marker, linewidth
x = np.arange(1, 8)

plt.figure(figsize=(10, 5))
plt.plot(x, x,       color='steelblue', marker='o', label="marker='o'")
plt.plot(x, x + 2,   color='indianred', linestyle='--', marker='s', label="linestyle='--'")
plt.plot(x, x + 4,   color='seagreen', linestyle=':', linewidth=3, label="linewidth=3")
plt.title('선 스타일 비교')
plt.legend()
plt.show()
````

## 2. 주요 차트 종류 — 언제 무엇을 쓸까?

| 차트 | 함수 | 언제 |
|------|------|------|
| 선 그래프 | `plt.plot` | 시간/순서에 따른 **변화** |
| 막대 그래프 | `plt.bar` | **범주 간 비교** |
| 히스토그램 | `plt.hist` | 숫자 하나의 **분포** |
| 산점도 | `plt.scatter` | 두 숫자 변수의 **관계** |

````python
# 막대 그래프 — 범주 비교
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 32]

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].bar(categories, values, color='skyblue', edgecolor='gray')
axes[0].set_title('세로 막대 (bar)')

# 범주 이름이 길면 가로 막대(barh)가 읽기 좋음
axes[1].barh(categories, values, color='lightcoral', edgecolor='gray')
axes[1].set_title('가로 막대 (barh)')
plt.tight_layout()
plt.show()
````

````python
# 히스토그램 — bins(구간 수)에 따라 보이는 모양이 달라짐!
data = np.random.randn(1000)

fig, axes = plt.subplots(1, 3, figsize=(13, 3.5))
for ax, bins in zip(axes, [5, 30, 100]):
    ax.hist(data, bins=bins, edgecolor='black', alpha=0.7)
    ax.set_title(f'bins={bins}')
plt.suptitle('같은 데이터, 다른 bins — 너무 적으면 뭉개지고, 너무 많으면 노이즈')
plt.tight_layout()
plt.show()
````

````python
# 산점도 — 두 변수의 관계. 색(c)과 크기(s)로 추가 정보 표현 가능
np.random.seed(0)
x = np.random.rand(100)
y = x * 2 + np.random.randn(100) * 0.2
colors = y - x * 2          # 잔차를 색으로
sizes = np.random.rand(100) * 100 + 20

plt.figure(figsize=(8, 5))
sc = plt.scatter(x, y, c=colors, s=sizes, alpha=0.6, cmap='coolwarm')
plt.colorbar(sc, label='잔차')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('산점도 — 양의 상관관계')
plt.show()
````

## 3. 서브플롯과 저장

여러 차트를 한 그림에 배치할 때는 `plt.subplots(행, 열)`을 씁니다.

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0, 0].plot(...)     # 왼쪽 위
axes[1, 1].hist(...)     # 오른쪽 아래
```

- Axes 객체에서는 `set_title()`, `set_xlabel()`처럼 `set_`이 붙습니다 (`plt.title()`과 다름!)
- `plt.tight_layout()`: 차트끼리 겹치지 않게 간격 자동 조정
- `plt.savefig('파일명.png', dpi=150)`: 파일로 저장 — **`show()` 전에** 호출

````python
np.random.seed(3)
fig, axes = plt.subplots(2, 2, figsize=(11, 7))

axes[0, 0].plot(np.cumsum(np.random.randn(100)))
axes[0, 0].set_title('선 그래프 (누적합)')

axes[0, 1].bar(['A', 'B', 'C'], [10, 20, 15], color='skyblue')
axes[0, 1].set_title('막대 그래프')

axes[1, 0].hist(np.random.randn(500), bins=25, alpha=0.7)
axes[1, 0].set_title('히스토그램')

axes[1, 1].scatter(np.random.rand(50), np.random.rand(50), alpha=0.6)
axes[1, 1].set_title('산점도')

plt.tight_layout()
plt.savefig('my_charts.png', dpi=150)   # 파일로도 저장
plt.show()
print("my_charts.png 저장 완료")
````

## 4. Seaborn 통계 시각화

Seaborn은 **DataFrame과 궁합이 좋습니다**. `data=df, x='열이름', y='열이름'` 패턴으로 통계 차트를 한 줄에 그립니다.

실습에는 Seaborn 내장 `tips` 데이터셋(식당 계산서 244건)을 사용합니다.

| 열 | 의미 |
|----|------|
| `total_bill` | 총 지출액 ($) |
| `tip` | 팁 ($) |
| `sex`, `smoker` | 성별, 흡연 여부 |
| `day`, `time` | 요일, 점심/저녁 |
| `size` | 일행 수 |

````python
tips = sns.load_dataset('tips')   # 최초 실행 시 인터넷에서 다운로드 후 캐시됨
print(tips.shape)
tips.head()
````

````python
# countplot: 범주별 개수 / barplot: 범주별 평균 (검은 선은 신뢰구간)
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

sns.countplot(data=tips, x='day', hue='day', palette='Set2', legend=False, ax=axes[0])
axes[0].set_title('요일별 방문 건수 (countplot)')

sns.barplot(data=tips, x='day', y='total_bill', hue='day', palette='Set2', legend=False, ax=axes[1])
axes[1].set_title('요일별 평균 지출 (barplot)')

plt.tight_layout()
plt.show()
````

````python
# boxplot: 분포를 5개 숫자(최소, 1분위, 중앙값, 3분위, 최대)로 요약. 점은 이상치!
# violinplot: boxplot + 분포 모양까지
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

sns.boxplot(data=tips, x='day', y='total_bill', hue='day', palette='Set3', legend=False, ax=axes[0])
axes[0].set_title('boxplot — 위쪽 점들이 이상치')

sns.violinplot(data=tips, x='day', y='total_bill', hue='day', palette='Set3', legend=False, ax=axes[1])
axes[1].set_title('violinplot — 분포 모양까지 보임')

plt.tight_layout()
plt.show()
````

````python
# histplot + kde: 분포 확인 / hue로 그룹 비교
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

sns.histplot(data=tips, x='total_bill', kde=True, bins=20, ax=axes[0])
axes[0].set_title('총 지출 분포')

sns.histplot(data=tips, x='total_bill', hue='time', kde=True, bins=20, ax=axes[1])
axes[1].set_title('점심 vs 저녁 — hue로 그룹 비교')

plt.tight_layout()
plt.show()
````

````python
# scatterplot: hue(색), size(크기)로 3~4개 변수를 한 차트에
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', size='size', alpha=0.7)
plt.title('지출 vs 팁 (색=시간대, 크기=일행 수)')
plt.show()
````

````python
# heatmap: 상관계수 행렬 시각화 — ML 특성 선택의 필수 도구
# 상관계수: -1(완전 반비례) ~ 0(무관) ~ +1(완전 비례)
corr = tips[['total_bill', 'tip', 'size']].corr()

plt.figure(figsize=(6, 4.5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, vmin=-1, vmax=1)
plt.title('상관관계 히트맵')
plt.show()
print("total_bill과 tip의 상관계수가 0.68 → 지출이 클수록 팁도 큰 경향")
````

## 5. ML 실무 시각화

ML 프로젝트에서 반복적으로 그리게 되는 차트들입니다.

````python
# 패턴 1: 타깃별 특성 분포 — "이 특성이 분류에 도움이 될까?"를 눈으로 확인
# 두 분포가 많이 겹치면 그 특성만으로는 구분이 어려움
np.random.seed(42)
df = pd.DataFrame({
    'feature1': np.concatenate([np.random.randn(50), np.random.randn(50) + 0.3]),   # 겹침 많음
    'feature2': np.concatenate([np.random.randn(50), np.random.randn(50) + 3]),     # 잘 분리됨
    'target': [0] * 50 + [1] * 50
})

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(data=df, x='feature1', hue='target', kde=True, ax=axes[0])
axes[0].set_title('feature1 — 겹침 많음 → 구분에 큰 도움 안 됨')
sns.histplot(data=df, x='feature2', hue='target', kde=True, ax=axes[1])
axes[1].set_title('feature2 — 잘 분리됨 → 좋은 특성!')
plt.tight_layout()
plt.show()
````

````python
# 패턴 2: 학습 곡선 — 훈련 데이터가 늘어날 때 점수 변화로 과적합/과소적합 진단
# 훈련 점수와 검증 점수의 간격이 크면 과적합 신호
from sklearn.model_selection import learning_curve
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification

X, y = make_classification(n_samples=300, n_features=10, random_state=42)

train_sizes, train_scores, val_scores = learning_curve(
    LogisticRegression(max_iter=1000), X, y, cv=5,
    train_sizes=np.linspace(0.1, 1.0, 10)
)

plt.figure(figsize=(9, 4.5))
plt.plot(train_sizes, train_scores.mean(axis=1), marker='o', label='훈련 점수')
plt.plot(train_sizes, val_scores.mean(axis=1), marker='s', label='검증 점수')
plt.xlabel('훈련 샘플 수')
plt.ylabel('정확도')
plt.title('학습 곡선 — 두 선의 간격이 좁아지면 데이터가 충분하다는 뜻')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

````python
# 패턴 3: 이미지 데이터 격자로 확인 — 이미지 ML의 첫 단계
# sklearn 내장 digits 데이터셋 (8x8 손글씨 숫자)
from sklearn.datasets import load_digits

digits = load_digits()
print(f"이미지 개수: {len(digits.images)}, 한 장 크기: {digits.images[0].shape}")

fig, axes = plt.subplots(3, 5, figsize=(8, 5))
for ax, img, label in zip(axes.ravel(), digits.images, digits.target):
    ax.imshow(img, cmap='gray_r')       # imshow: 2차원 배열을 이미지로
    ax.set_title(f'라벨: {label}', fontsize=9)
    ax.axis('off')                      # 눈금 숨기기
plt.tight_layout()
plt.show()
````

|------|-----------|

| ML 시각화 | 타깃별 특성 분포, 학습 곡선, `imshow` 이미지 격자 |
