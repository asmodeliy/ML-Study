```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# 한글 폰트 설정 (Windows)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

## 1. Matplotlib 기본

```python
# 선 그래프
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(10, 5))
plt.plot(x, y, label='sin(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('사인 함수')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 막대 그래프
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 32]

plt.figure(figsize=(8, 5))
plt.bar(categories, values, color='skyblue')
plt.xlabel('카테고리')
plt.ylabel('값')
plt.title('막대 그래프')
plt.show()
```

```python
# 히스토그램
data = np.random.randn(1000)

plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, edgecolor='black', alpha=0.7)
plt.xlabel('값')
plt.ylabel('빈도')
plt.title('히스토그램')
plt.show()
```

```python
# 산점도
x = np.random.rand(100)
y = x * 2 + np.random.randn(100) * 0.1

plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.6)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('산점도')
plt.show()
```

```python
# 서브플롯
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

axes[0, 0].plot(np.random.rand(10))
axes[0, 0].set_title('선 그래프')

axes[0, 1].bar(['A', 'B', 'C'], [10, 20, 15])
axes[0, 1].set_title('막대 그래프')

axes[1, 0].hist(np.random.randn(100))
axes[1, 0].set_title('히스토그램')

axes[1, 1].scatter(np.random.rand(50), np.random.rand(50))
axes[1, 1].set_title('산점도')

plt.tight_layout()
plt.show()
```

## 2. Seaborn 통계 시각화

```python
# 샘플 데이터셋
tips = sns.load_dataset('tips')
print(tips.head())
```

```python
# 막대 그래프
plt.figure(figsize=(8, 5))
sns.barplot(x='day', y='total_bill', data=tips, palette='Set2')
plt.title('요일별 평균 총 지출')
plt.show()
```

```python
# 박스 플롯
plt.figure(figsize=(8, 5))
sns.boxplot(x='day', y='total_bill', data=tips, palette='Set3')
plt.title('요일별 총 지출 분포')
plt.show()
```

```python
# 히스토그램 ( KDE )
plt.figure(figsize=(8, 5))
sns.histplot(tips['total_bill'], kde=True, bins=20)
plt.title('총 지출 분포')
plt.show()
```

```python
# 상관관계 히트맵
plt.figure(figsize=(8, 6))
corr = tips[['total_bill', 'tip', 'size']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.title('상관관계 히트맵')
plt.show()
```

```python
# 조인트 플롯
sns.jointplot(x='total_bill', y='tip', data=tips, kind='reg')
plt.suptitle('총 지출 vs 팁', y=1.02)
plt.show()
```

## 3. ML 실무 시각화

```python
# 특성 분포 확인 (ML 전처리에 유용)
np.random.seed(42)
df = pd.DataFrame({
    'feature1': np.random.randn(100),
    'feature2': np.random.randn(100) * 2 + 5,
    'target': np.random.choice([0, 1], 100)
})

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(data=df, x='feature1', hue='target', kde=True, ax=axes[0])
axes[0].set_title('Feature 1 분포')

sns.histplot(data=df, x='feature2', hue='target', kde=True, ax=axes[1])
axes[1].set_title('Feature 2 분포')

plt.tight_layout()
plt.show()
```

```python
# 학습 곡선 시각화 (ML 모델 평가에 유용)
from sklearn.model_selection import learning_curve
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification

X, y = make_classification(n_samples=100, n_features=10, random_state=42)

train_sizes, train_scores, val_scores = learning_curve(
    LogisticRegression(), X, y, cv=5, n_jobs=-1,
    train_sizes=np.linspace(0.1, 1.0, 10)
)

plt.figure(figsize=(10, 5))
plt.plot(train_sizes, train_scores.mean(axis=1), label='훈련 점수')
plt.plot(train_sizes, val_scores.mean(axis=1), label='검증 점수')
plt.xlabel('훈련 샘플 수')
plt.ylabel('점수')
plt.title('학습 곡선')
plt.legend()
plt.grid(True)
plt.show()
```
